# Example for registering files to the DESC Data Registry

```python
owner = "YOUR_NERSC_USERNAME_HERE"

import dataregistry
print(f"Working with dataregistry version: {dataregistry.__version__} as owner {owner}")
```

```python
datareg = dataregistry.DataRegistry(schema = "lsst_desc_working",
                                    owner = owner)
```

```python
id_dataset, id_execution = datareg.register_dataset(
    name            = "2pt_NG_final_2ptunblind_02_26_21_wnz_maglim_covupdate",
    version         = "1.0.1",
    description     = "2pt_NG_final_2ptunblind_02_26_21_wnz_maglim_covupdate from DES Y3 (https://des.ncsa.illinois.edu/releases/y3a2/Y3key-products)",
    owner_type      = "user",
    is_overwritable = True,
    old_location    = '/global/homes/s/sajkov/DEED/example_analysis_files/2pt_NG_final_2ptunblind_02_26_21_wnz_maglim_covupdate.fits'
)

print(f"Dataset {id_dataset} created.")
print(f"Dataset assigned to execution {id_execution}.")
```
